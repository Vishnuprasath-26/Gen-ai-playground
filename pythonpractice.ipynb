```python
#question 1
#Create a Python program using a class Employee to calculate and display an employee's annual salary and determine whether it exceeds ₹8,00,000.
#Requirements:
#Create an Employee class.
#Initialize the employee's name and monthly salary using a constructor.
#Create a display() method to:
#Display the employee name.
#Display the monthly salary.
#Calculate and display the annual salary.
#Check whether the annual salary exceeds ₹8,00,000.
#Get the employee name and monthly salary from the user.
#Create an Employee object and call the display() method

class Employee:
    def __init__(self, name, monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary
    def display(self):
        annual_salary = self.monthly_salary * 12
        print("Employee Name :", self.name)
        print("Monthly Salary :", self.monthly_salary)
        print("Annual Salary :", annual_salary)
        if annual_salary > 800000:
            print("Salary exceeds ₹8,00,000")
        else:
            print("Salary does not exceed ₹8,00,000")
name = input("Enter Employee Name : ")
salary = int(input("Enter Monthly Salary : "))
emp = Employee(name, salary)
emp.display()
    
```

```python
#question 2
# Write a Python program to store student names and marks, calculate the average marks,
# display students who scored above the average, and find the topper using a list of
# tuples and a dictionary.

# Requirements:
#Create an empty list called students.
#Get the number of students from the user.
#For each student:
# - Get the student name.
# - Get the student's marks.
# - Store the name and marks as a tuple inside the list.
#Calculate the total marks and average marks of all students.
#Display the average marks.
#Display the students whose marks are above the average.
#Find the student who scored the highest marks (Topper).
#Store the topper's name and marks in a dictionary.
#Display the topper details.

students = []
n = int(input("Enter Number of Students : "))
for i in range(n):
    name = input("Enter Student Name : ")
    marks = int(input("Enter Student Marks : "))
    students.append((name, marks))
topper = students[0]
total = 0
for student in students:
    total = total + student[1]
    if student[1] > topper[1]:
        topper = student
average = total / n
print("Average Marks :", average)
print("Students Above Average")
for student in students:
    if student[1] > average:
        print(student)
topper_dict = {}
topper_dict["Name"] = topper[0]
topper_dict["Marks"] = topper[1]
print("Topper Details")
print(topper_dict)
```

```python
#question 3
#Write a Python program to analyze product sales using a dictionary and generate a sales summary for each product.
#Requirements:
#Create a dictionary named sales containing product names as keys and a list of sales amounts as values.
#Calculate the total sales for each product.
#Calculate the average sales for each product.
#Store the total and average sales of each product in a new dictionary called summary.
#Find the highest-selling product based on its total sales.
#Display:
#Each product's total sales.
#Each product's average sales.
#The highest-selling product.
#The highest sales amount.
#The complete summary dictionary.

sales = {
    "Laptop": [50000, 45000, 60000],
    "Mobile": [30000, 35000, 40000],
    "Tablet": [20000, 25000, 22000]
}
summary = {}
highest_product = ""
highest_total = 0
for product in sales:
    total = 0
    for amount in sales[product]:
        total = total + amount
    average = total / len(sales[product])
    summary[product] = {
        "Total": total,
        "Average": average
    }
    if total > highest_total:
        highest_total = total
        highest_product = product
print("Product Summary")
for product in summary:
    print(product)
    print("Total Sales :", summary[product]["Total"])
    print("Average Sales :", summary[product]["Average"])
    print()
print("Highest Selling Product :", highest_product)
print("Highest Sales :", highest_total)
print("Summary Dictionary")
print(summary)
```

```python
#question 4
#Write a Python program using a class Bank to perform basic banking operations such as deposit, withdrawal, and balance checking.
#Requirements:
#Create a class named Bank.
#Initialize the account balance to 0 using a constructor.
#Create a deposit() method to:
#Get the deposit amount from the user.
#Check whether the amount is greater than 0.
#Add the valid amount to the balance.
#Display an appropriate message for invalid amounts.
#Create a withdraw() method to:
#Get the withdrawal amount from the user.
#Check whether sufficient balance is available.
#Deduct the amount if sufficient balance exists.
#Display "Insufficient Balance" otherwise.
#Create a balance_check() method to display the current balance.
#Create an object of the Bank class and perform deposit, withdrawal, and balance-check operations.

class Bank:
    def __init__(self):
        self.balance = 0
    def deposit(self):
        amount = int(input("Enter Deposit Amount : "))
        if amount > 0:
            self.balance = self.balance + amount
            print("Money Deposited")
        else:
            print("Invalid Amount")
    def withdraw(self):
        amount = int(input("Enter Withdraw Amount : "))
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Money Withdrawn")
        else:
            print("Insufficient Balance")
    def balance_check(self):
        print("Balance :", self.balance)
b = Bank()
b.deposit()
b.withdraw()
b.balance_check()
```

```python
#5.Find the Second Largest Number in a List
#Write a Python program to find the second largest number without using sort().
numbers = [10, 45, 20, 80, 35, 60]
largest = numbers[0]
second_largest = numbers[0]
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num
print("Largest:", largest)
print("Second Largest:", second_largest)
```

```python
#6.Find Duplicate Elements and Their Count
#Write a Python program to find duplicate elements and count how many times each appears.

numbers = [10, 20, 30, 20, 40, 10, 20, 50, 30]
count = {}
for num in numbers:
    if num in count:
        count[num] += 1
    else:
        count[num] = 1
print("Duplicate elements:")
for num in count:
    if count[num] > 1:
        print(num, "appears", count[num], "times")
```
