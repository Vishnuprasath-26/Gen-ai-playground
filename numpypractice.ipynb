```python
#Create a NumPy array from 1 to 20.

import numpy  as np 
arr=np.arange(1,21)
print(arr)
```

```python
#Print only even numbers.
arr=np.arange(1,21)
print (arr[arr%2==0])
```

```python
#Print only odd numbers.
arr=np.arange(1,21)
print(arr[arr%2!=0])
```

```python
#Find the sum of all elements.
arr=np.array([10,20,30,40,50])
print (arr.sum())
```

```python
#Find the average
arr=np.array([10,20,30,40,50])
print(arr.mean())
```

```python
#Find the average
arr=np.array([10,20,30,40,50])
print("maximum",arr.max())
print("minimum",arr.min())
```

```python
#Reverse the array
arr=np.array([10,20,30,40,50])
print(arr[::-1])
```

```python
#Convert integer array into float array.
arr=np.array([1,2,3,4,5])
arr1=arr.astype(float)
print(arr1)
print(arr1.dtype)
```

```python
#Reshape a 1D array into a 3×4 matrix.
arr=np.arange(1,13)
matrix=arr.reshape(3,4)
print(matrix)
```

```python
#Flatten a 2D array into a 1D array.
arr=np.array([[1,2,3],
             [4,5,6]])
flat=arr.flatten()
print(flat)
```
