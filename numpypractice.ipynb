```python
#Find the row with the highest sum
arr=np.array([[10,20,30],[40,50,60],[70,80,90]])
rowsum=np.sum(arr,axis=1)
maxrow=np.argmax(rowsum)
print("row sum:",rowsum)
print("highest sum row index",maxrow)
print("row:",arr[maxrow])
```

```python
#Find the column with the highest mean
arr=np.array([[10,20,30],[40,50,60],[70,80,90]])
meancolm=np.mean(arr,axis=0)
maxcolm=np.argmax(meancolm)
print("column means:",meancolm)
print("highest mean column index:",maxcolm)
print("column:",arr[:,maxcolm])
```

```python
#Find common values between two arrays
a=np.array([1,2,3,4,5])
b=np.array([1,2,3,4,5])
result=np.intersect1d(a,b)
print("comman values:",result)
```

```python
#Find the Top 3 Values
arr=np.array([10,70,50,30,80])
sort=np.sort(arr)
top3=sort[-3:][::-1]
print("top 3 values:",top3)
```

```python
#Find the position of the maximum value
arr = np.array([[10, 20, 30],[90, 40, 50],[60, 70, 80]])
flat_index = np.argmax(arr)
row, column = np.unravel_index(flat_index, arr.shape)
print("Maximum value:", arr[row, column])
print("Row:", row)
print("Column:", column)
```

```python
#find the maximum value from the each rows
arr=np.array([[10,50,30],[80,20,30],[60,90,70]])
result=np.max(arr,axis=1)
print(result)
```

```python
#Find column-wise minimum values
arr=np.array([[10,50,30],[80,20,30],[60,90,70]])
result=np.min(arr,axis=0)
print(result)
```

```python
#Find values that appear more than once
arr=np.array([1,2,3,2,4,5,3,6])
values, counts = np.unique(arr, return_counts=True)
result = values[counts > 1]
print(result)
```
