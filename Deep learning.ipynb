```python

import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# =========================================================
# 1. LOAD DATASET
# =========================================================

df = pd.read_csv("BankNote_Authentication.csv")

print("Original Dataset:")
display(df.head())

print("\nDataset Shape:", df.shape)


# =========================================================
# 2. DATA CLEANING
# =========================================================

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nCleaned Dataset:")
display(df.head())

print("\nCleaned Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)


# =========================================================
# 3. FEATURES AND LABELS
# =========================================================

X = df[
    [
        "variance",
        "skewness",
        "curtosis",
        "entropy"
    ]
].values

y = df["class"].values

print("\nFeatures Shape:", X.shape)
print("Labels Shape:", y.shape)


# =========================================================
# 4. STEP FUNCTION
# =========================================================

def step_function(x):
    return 1 if x >= 0 else 0


# =========================================================
# 5. PERCEPTRON CLASS
# =========================================================

class Perceptron:

    def __init__(self, input_size):
        self.weights = np.zeros(input_size)
        self.bias = 0
        self.learning_rate = 0.15

    def predict(self, x):
        weighted_sum = np.dot(self.weights, x) + self.bias
        return step_function(weighted_sum)

    def train(self, X, y, epochs):

        for epoch in range(epochs):

            for xi, yi in zip(X, y):

                y_pred = self.predict(xi)

                error = yi - y_pred

                self.weights += (
                    self.learning_rate * error * xi
                )

                self.bias += (
                    self.learning_rate * error
                )


# =========================================================
# 6. CREATE PERCEPTRON
# =========================================================

perceptron = Perceptron(input_size=4)

print("\nInitial Weights:", perceptron.weights)
print("Initial Bias:", perceptron.bias)


# =========================================================
# 7. TRAIN MODEL
# =========================================================

perceptron.train(X, y, epochs=300)

print("\nModel Training Completed")

print("Final Weights:", perceptron.weights)
print("Final Bias:", perceptron.bias)


# =========================================================
# 8. FIRST 10 PREDICTIONS
# =========================================================

print("\nFirst 10 Predictions:")

for xi in X[:10]:

    prediction = perceptron.predict(xi)

    print(
        f"Input: {xi}, "
        f"Prediction: {prediction}"
    )


# =========================================================
# 9. PREDICT COMPLETE DATASET
# =========================================================

y_pred = np.array([
    perceptron.predict(xi)
    for xi in X
])


# =========================================================
# 10. CONFUSION MATRIX
# =========================================================

cm = confusion_matrix(y, y_pred)

print("\nConfusion Matrix:")
print(cm)


# =========================================================
# 11. EVALUATION METRICS
# =========================================================

accuracy = accuracy_score(y, y_pred)

precision = precision_score(y, y_pred)

recall = recall_score(y, y_pred)

f1 = f1_score(y, y_pred)


print("\nEvaluation Metrics")
print("-------------------")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)


# =========================================================
# 12. CLASSIFICATION REPORT
# =========================================================

print("\nClassification Report:")
print(classification_report(y, y_pred))

```
