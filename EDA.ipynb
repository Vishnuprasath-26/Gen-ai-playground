```python
#5.groupby()
#Group employees by Department and find the average salary of each department.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df.groupby("Department", as_index=False)["Salary"].mean()
print(result)
```

```python
#6.Aggregation Functions
#For each department, find:
#Minimum salary
#Maximum salary
#Average salary

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df.groupby("Department", as_index=False)["Salary"].agg(
    ["min", "max", "mean"]
)
print(result)
```

```python
#7.value_counts()
#Count how many employees belong to each department

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df["Department"].value_counts()
print(result)
```

```python
#8.unique()
#Find all unique departments available in the employee dataset.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df["Department"].unique()
print(result)
```

```python
#9.nunique()
#Find the total number of unique departments in the employee dataset.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df["Department"].nunique()
print(result)
```

```python
#10.sort_values()
#Sort the employees based on salary from highest to lowest.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df.sort_values("Salary", ascending=False)
print(result)
```

```python
#11.corr()
#Find the correlation between the numerical columns Age, Experience, and Salary. Identify which two variables have the strongest relationship.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Age": [25, 28, 30, 26, 29],
    "Experience": [2, 5, 7, 3, 6],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = df[["Age", "Experience", "Salary"]].corr()
print(result)
```

```python
#12.Pivot Table
#Create a pivot table that shows the average salary for each department.

data = {
    "Name": ["Arun", "Priya", "Raj", "Kumar", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 60000, 35000, 55000]
}
df = pd.DataFrame(data)
result = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="mean"
)
print(result)
```
