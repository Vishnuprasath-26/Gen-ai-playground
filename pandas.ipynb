```python
#1.Series
#Create a Pandas Series containing 5 employee salaries. Find the highest, lowest, and average salary.
import pandas as pd
salary=pd.Series([1000,2000,3000,4000,5000])
print("Employee salary")
print(salary)
print("Highest salary:",salary.max())
print("Lowest salary:",salary.min())
print("Average salary:",salary.mean())
```

```python
#2.DataFrame
#Create a DataFrame containing employee name, age, department, and salary. Display the complete DataFrame.
data={"Name":["vishnu","harini","ragul"],
      "age":[21,24,23],
      "salary":["it","hr","it"]}
employes=pd.DataFrame(data)
print(employes)
```

```python
#3.DataFrame using Dictionary
#Create an employee DataFrame using a dictionary containing 5 employees and their salaries. Find the employee with the highest salary.
employees = {
    "Arun": 30000,
    "Priya": 45000,
    "Ravi": 35000,
    "Meena": 50000,
    "Kumar": 40000
}
data = pd.DataFrame(
    employees.items(),
    columns=["Name", "Salary"]
)
highest = data["Salary"].max()
print(data)
print("Highest Salary :", highest)
print("Employee       :", data.loc[data["Salary"].idxmax(), "Name"])
```

```python
#4. DataFrame using List
#Create a student DataFrame using a list containing Name, Mark, and Department. Display students who scored above 80.
students=[["arun",85,"it"],
          ["priya",75,"cse"],
          ["ravi",90,"cse"]]
data=pd.DataFrame(students,columns=["name","mark","department"])
print(data)
print("student scored above 80:")
print(data[data["mark"]>80])

```

```python
#5.DataFrame using NumPy Array
#Create a NumPy array containing 10 employee salaries and convert it into a Pandas DataFrame. Find the average salary.
import numpy as np
import pandas as pd
salary=np.array([1000,2000,3000,4000,5000,6000,7000,8000,9000,10000])
data=pd.DataFrame(salary,columns=["salary"])
print(data)
print("average salary:",data["salary"].mean())
```

```python
#11.Add New Row
#Add a new employee to an existing employee DataFrame.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [22, 24, 23],
    "Department": ["IT", "HR", "IT"],
    "Salary": [30000, 35000, 40000]
})

new_employee = {
    "Name": "Meena",
    "Age": 25,
    "Department": "Finance",
    "Salary": 45000
}
data.loc[len(data)]=new_employee
print(data)

```

```python
#12.Add New column
#Add a Bonus column where every employee receives 10% of their salary.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Salary": [30000, 40000, 50000]
})

data["Bonus"] = data["Salary"] * 10 / 100

print(data)
```

```python
#13.Select One Column
#Select only the Salary column and find its average.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Salary": [30000, 40000, 50000]
})

salary = data["Salary"]

print("Salary:")
print(salary)

print("Average Salary :", salary.mean())
```

```python
#14.Select Multiple Columns
#From an employee DataFrame, display only Name, Department, and Salary.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [22, 24, 23],
    "Department": ["IT", "HR", "IT"],
    "Salary": [30000, 40000, 50000]
})

result = data[["Name", "Department", "Salary"]]

print(result)
```

```python
#15.Filtering
#Display employees whose salary is greater than ₹50,000.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi", "Meena", "Kumar"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 55000, 45000, 60000, 50000]
})

result = data[data["Salary"] > 50000]

print(result)
```

```python
#16.Update Column
#Increase every employee's salary by 10%.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Salary": [30000, 40000, 50000]
})

data["Salary"] = data["Salary"] * 1.10

print(data)
```

```python
#17.Delete Column
#Delete the Age column from an employee DataFrame
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [22, 24, 23],
    "Department": ["IT", "HR", "IT"],
    "Salary": [30000, 40000, 50000]
})

data = data.drop("Age", axis=1)

print(data)
```

```python
#18.Delete Rows
#Question: Delete the employee record at index 3.
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi", "Meena", "Kumar"],
    "Age": [22, 24, 23, 25, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 40000, 45000, 50000, 55000]
})

data = data.drop(3, axis=0)

print(data)
```
