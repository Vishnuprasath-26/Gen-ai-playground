```python
#1.What is Data Cleaning?
#You are given an employee dataset. Explain what Data Cleaning means and why it is needed before performing analysis.
import pandas as pd
data={
    "Name": ["Arun", "Priya", "Raj"],
    "Age": [25, 28, 30],
    "Salary": [30000, 45000, 60000]
}
df=pd.DataFrame(data)
print(df)
```

```python
#2.Copy Dataset
#Create a copy of an employee DataFrame, modify the copied DataFrame, and verify that the original DataFrame remains unchanged.
data = {
    "Name": ["Arun", "Priya", "Raj"],
    "Salary": [30000, 45000, 60000]
}

df = pd.DataFrame(data)
df_copy=df.copy()
df_copy["Salary"]=df_copy["Salary"]+5000

print(df)
print("copy dataframe")
print(df_copy)
```

```python
#3.Checking Missing Values
#Check whether the employee dataset contains any missing values.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, 50000]
}

df = pd.DataFrame(data)
print(df.isnull())
```

```python
#4.Count Missing Values
#Find the number of missing values in each column of the employee dataset.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, 50000]
}

df = pd.DataFrame(data)
print(df.isnull().sum())
```

```python
#5.Find Column with Most Missing Values
#Find which column contains the highest number of missing values in the employee dataset.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, None]
}
df = pd.DataFrame(data)
missing = df.isnull().sum()
print("Column with most missing values:", missing.idxmax())
print("Number of missing values:", missing.max())
```

```python
#6.Drop One Column with Missing Values
# A column named Email contains many missing values. Remove the entire Email column from the DataFrame.

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, 30, 26],
    "Salary": [30000, None, 60000, 50000],
    "Email": ["arun@gmail.com", None, None, "meena@gmail.com"]
}
df = pd.DataFrame(data)
df = df.drop("Email", axis=1)
print(df)
```
