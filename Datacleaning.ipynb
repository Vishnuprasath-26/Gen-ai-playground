```python
#1.What is Data Cleaning?
#You are given an employee dataset. Explain what Data Cleaning means and why it is needed before performing analysis.
import pandas as pd
data={
    "Name": ["Arun", "Priya", "Raj"],
    "Age": [25, 28, 30],
    "Salary": [30000, 45000, 60000]
}
df=pd.DataFrame(data)
print(df)
```

```python
#2.Copy Dataset
#Create a copy of an employee DataFrame, modify the copied DataFrame, and verify that the original DataFrame remains unchanged.
data = {
    "Name": ["Arun", "Priya", "Raj"],
    "Salary": [30000, 45000, 60000]
}

df = pd.DataFrame(data)
df_copy=df.copy()
df_copy["Salary"]=df_copy["Salary"]+5000

print(df)
print("copy dataframe")
print(df_copy)
```

```python
#3.Checking Missing Values
#Check whether the employee dataset contains any missing values.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, 50000]
}

df = pd.DataFrame(data)
print(df.isnull())
```

```python
#4.Count Missing Values
#Find the number of missing values in each column of the employee dataset.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, 50000]
}

df = pd.DataFrame(data)
print(df.isnull().sum())
```

```python
#5.Find Column with Most Missing Values
#Find which column contains the highest number of missing values in the employee dataset.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, None, 30],
    "Salary": [30000, None, 60000, None]
}
df = pd.DataFrame(data)
missing = df.isnull().sum()
print("Column with most missing values:", missing.idxmax())
print("Number of missing values:", missing.max())
```

```python
#6.Drop One Column with Missing Values
# A column named Email contains many missing values. Remove the entire Email column from the DataFrame.

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Age": [25, 28, 30, 26],
    "Salary": [30000, None, 60000, 50000],
    "Email": ["arun@gmail.com", None, None, "meena@gmail.com"]
}
df = pd.DataFrame(data)
df = df.drop("Email", axis=1)
print(df)
```

```python
#7.Fill Missing Values
#The Salary column contains missing values. Fill the missing salary values with the mean salary.

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Salary": [30000, None, 60000, 50000]
}
df = pd.DataFrame(data)
mean_salary = df["Salary"].mean()
df["Salary"] = df["Salary"].fillna(mean_salary)
print(df)
```

```python
#8.Check Missing Values After Filling
#After filling the missing salary values, verify that no missing values remain in the Salary column.

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Salary": [30000, None, 60000, 50000]
}
df = pd.DataFrame(data)
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df["Salary"].isnull().sum())
```

```python
#9.Find Most Frequent Value
#The Department column contains missing values. Find the most frequently occurring department

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena", "Kumar"],
    "Department": ["IT", "HR", "IT", None, "IT"]
}
df = pd.DataFrame(data)
most_frequent = df["Department"].mode()[0]
print("Most frequent department:", most_frequent)
```

```python
#10.Fill Missing Values Using Most Frequent Value
#The Department column contains missing values. Fill them using the most frequently occurring department

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena", "Kumar"],
    "Department": ["IT", "HR", "IT", None, "IT"]
}
df = pd.DataFrame(data)
most_frequent = df["Department"].mode()[0]
df["Department"] = df["Department"].fillna(most_frequent)
print(df)
```

```python
#11.Mean
#Find the average salary of all employees and use it to fill any missing salary values.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Salary": [30000, None, 60000, 50000]
}
df = pd.DataFrame(data)
mean_salary = df["Salary"].mean()
print("Mean Salary:", mean_salary)
df["Salary"] = df["Salary"].fillna(mean_salary)
print(df)
```

```python
#12.Median
#Find the median salary of all employees and use it to fill any missing salary values.
data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Salary": [30000, None, 60000, 50000]
}
df = pd.DataFrame(data)
median_salary = df["Salary"].median()
print("Median Salary:", median_salary)
df["Salary"] = df["Salary"].fillna(median_salary)
print(df)
```

```python
#13.apply()
#apply() is used to apply a function to each value in a Pandas column.

df = pd.DataFrame({
    "Student": ["Arun", "Bala", "Kumar", "Ravi", "Vijay"],
    "Marks": [95, 82, 67, 45, 75]
})
def grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 50:
        return "C"
    else:
        return "D"
df["Grade"] = df["Marks"].apply(grade)
print(df)
```

```python
#14.map()
#map() is commonly used to replace values in a Pandas Series based on a dictionary.

df = pd.DataFrame({
    "Student": ["Arun", "Bala", "Kumar", "Ravi"],
    "Department": ["IT", "HR", "FN", "IT"]
})
department_map = {
    "IT": "Information Technology",
    "HR": "Human Resources",
    "FN": "Finance"
}
df["Department"] = df["Department"].map(department_map)
print(df)
```

```python
#15.Lambda
#Increase every employee's salary by 10% using apply() and a lambda function.

data = {
    "Name": ["Arun", "Priya", "Raj", "Meena"],
    "Salary": [30000, 45000, 60000, 50000]
}
df = pd.DataFrame(data)
df["Salary"] = df["Salary"].apply(lambda x: x * 1.10)
print(df)
```

```python
#16.String/Text Cleaning.  
#Clean Name and City by removing spaces and converting text to lowercase.

df = pd.DataFrame({
    "Name": ["  VISHNU  ", " ARUN", "Priya  "],
    "City": [" Chennai ", "CHENNAI", " Madurai "]
})

df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()

df["Name"] = df["Name"].str.lower()
df["City"] = df["City"].str.lower()

print(df)
```
